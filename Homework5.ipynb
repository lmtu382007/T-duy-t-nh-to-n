```python
# W5A16: Viết hàm kiểm tra một số có phải số Armstrong hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x
def is_armstrong(x):
    str_x = str(x)
    num_digits = len(str_x)
    armstrong_sum = sum(int(digit) ** num_digits for digit in str_x)
    return armstrong_sum == x
print(is_armstrong(153))  
print(is_armstrong(9474)) 
print(is_armstrong(123))  
```

```python
# W5A17: Viết hàm kiểm tra một số có phải số chính phương hay không? Hàm trả về kiểu bool, có tham số là một số nguyên x
def is_perfect_square(x):
    if x < 0:
        return False
    root = int(x ** 0.5)
    return root * root == x
print(is_perfect_square(16))
print(is_perfect_square(20))
print(is_perfect_square(-4))
```

```python
# W5A18: Viết hàm kiểm tra một số có phải số đối xứng không, xử lý trên dữ liệu kiểu số nguyên.
def is_palindrome_number(x):
    str_x = str(x)
    return str_x == str_x[::-1]
print(is_palindrome_number(121))
print(is_palindrome_number(-121))
print(is_palindrome_number(10))
```

```python
# W5A19: Viết hàm kiểm tra một số có phải số kỳ diệu hay không, xử lý trên dữ liệu kiểu số nguyên.
def is_magic_number(x):
    def sum_of_digits(n):
        return sum(int(digit) for digit in str(n))
    
    result = x
    while result >= 10:
        result = sum_of_digits(result)
    
    return result == 1
print(is_magic_number(19))
print(is_magic_number(123))
```

```python
# W5A20: Viết hàm detect_number_type(s) trả về 'int' nếu s là số nguyên, 'float' nếu là số thực, ngược lại 'neither'.
def detect_number_type(s):
    try:
        int(s)
        return 'int'
    except ValueError:
        try:
            float(s)
            return 'float'
        except ValueError:
            return 'neither'
print(detect_number_type("123"))
print(detect_number_type("123.45"))
```

```python
# W5A23: Viết hàm quadratic_roots_count(a, b, c) trả về số nghiệm thực của phương trình a*x^2 + b*x + c = 0.
def quadratic_roots_count(a, b, c):
    discriminant = b**2 - 4*a*c
    if discriminant > 0:
        return 2
    elif discriminant == 0:
        return 1
    else:
        return 0
print(quadratic_roots_count(1, -3, 2))
print(quadratic_roots_count(1, -2, 1))
```

```python
# W5A24: Viết hàm are_collinear(x1,y1,x2,y2,x3,y3) kiểm tra ba điểm có thẳng hàng hay không.
def are_collinear(x1, y1, x2, y2, x3, y3):
    return (y2 - y1) * (x3 - x2) == (y3 - y2) * (x2 - x1)
print(are_collinear(1, 2, 2, 3, 3, 4))
print(are_collinear(1, 2, 2, 3, 3, 5))
```

```python
# W5A25: Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác quad hay không. Trong đó: quad là một tứ giác (4 đỉnh – 4 điểm), p là một điểm cần kiểm tra. VD: quad = [(0,0), (4,0), (4,3), (0,3)], p = (2,1)
def point_in_quadrilateral(quad, p):
    def area(x1, y1, x2, y2, x3, y3):
        return abs((x1*(y2 - y3) + x2*(y3 - y1) + x3*(y1 - y2)) / 2.0)
    
    A = area(*quad[0], *quad[1], *quad[2]) + area(*quad[0], *quad[2], *quad[3])
    A1 = area(*p, *quad[0], *quad[1])
    A2 = area(*p, *quad[1], *quad[2])
    A3 = area(*p, *quad[2], *quad[3])
    A4 = area(*p, *quad[3], *quad[0])
    
    return A == A1 + A2 + A3 + A4
quad = [(0,0), (4,0), (4,3), (0,3)]
p = (2,1)
print(point_in_quadrilateral(quad, p))
p = (5,1)
print(point_in_quadrilateral(quad, p))
```

```python
# W5A26: Viết hàm trả về a XOR b.
def xor(a, b):
    return (a and not b) or (not a and b)
print(xor(True, False))
print(xor(True, True))
print(xor(False, False))
```

```python
# W5A27: Viết hàm trả về số lượng chữ số khác 0 trong n số nguyên nhập vào từ bàn phím (một list)
def count_non_zero_digits(numbers):
    count = 0
    for number in numbers:
        count += sum(1 for digit in str(abs(number)) if digit != '0')
    return count
numbers = [102, 305, -400, 0]
print(count_non_zero_digits(numbers))
```

```python
# W5A28: Viết hàm tính góc ∠ABC (độ) từ tọa độ A(x1,y1), B(x2,y2), C(x3,y3).
import math
def angle_abc(x1, y1, x2, y2, x3, y3):
    ab = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
    bc = math.sqrt((x3 - x2) ** 2 + (y3 - y2) ** 2)
    ac = math.sqrt((x3 - x1) ** 2 + (y3 - y1) ** 2)
    if ab == 0 or bc == 0:
        return None
    cos= (ab**2 + bc**2 - ac**2) / (2 * ab * bc)
    rad = math.acos(cos)
    deg = math.degrees(rad)
    return deg
print(angle_abc(0, 0, 1, 0, 1, 1))
print(angle_abc(0, 0, 1, 0, 2, 0))
```

```python
# W5A33: Viết hàm smart_taxi_fare(car_type, start_hour, km, wait_minutes) tính tổng tiền taxi theo quy tắc giá mở cửa, giá km, phí chờ, phụ phí ban đêm và giảm giá.
def smart_taxi_fare(car_type, start_hour, km, wait_minutes):
    if car_type == 'A':
        giagoc = 15000
        socay = 12000
    elif car_type == 'B':
        giagoc = 12000
        socay = 10000
    else:
        return "Loại xe không hợp lệ"
    
    gia = giagoc + km * socay + wait_minutes * 2000
    
    if start_hour >= 22 or start_hour < 6:
        gia *= 1.2
    
    if km > 30:
        gia *= 0.9
    
    return gia
print(smart_taxi_fare('A', 23, 35, 10))
```

```python
# W5A34: Viết hàm sum_1_to_n(n) trả về tổng các số x từ 1 đến n, nếu x có các chữ số đôi một khác nhau. Nếu n ≤ 0 trả về 0.
def sum_1_to_n(n):
    if n <= 0:
        return 0
    total = 0
    for x in range(1, n + 1):
        str_x = str(x)
        if len(set(str_x)) == len(str_x):
            total += x
    return total
print(sum_1_to_n(20))
print(sum_1_to_n(100))
```
