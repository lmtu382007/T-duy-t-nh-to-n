```python
a=int(input("Nhập vào một số nguyên dương, "))
S=0
for i in range(1, a):
    if a%i==0:
        S+=i
if S==a:
    print(a, "là số hoàn hảo")
else:
    print(a, "không phải là số hoàn hảo")
```

```python
##### Bài 12: Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n 
n=int(input("Nhập vào một số nguyên dương, "))
k=0
while (k*(k+1))//2 < n:
    k+=1
```

##### Bài 13: Nhập vào số nguyên dương A, tìm n nhỏ nhất sao cho:
##### 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 


```python
A=float(input("Nhập vào một số nguyên dương, "))
n=0
S=0
while S<=A:
    n+=1
    S+=1/n
print("n nhỏ nhất là:", n)
```

##### Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng 

```python
max_num=None
min_num=None
while True:
    num = int(input("Nhập vào một số nguyên (nhập -1 để dừng): "))
    if num == -1:
        break
    if max_num is None or num > max_num:
        max_num = num
    if min_num is None or num < min_num:
        min_num = num
print("Số lớn nhất trong các số vừa nhập là:", max_num)
print("Số nhỏ nhất trong các số vừa nhập là:", min_num)
```

##### Bài 15: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số. Làm bằng nhiều cách nhất có thể, ko dùng mảng. 


```python
n=int(input("Nhập vào một số nguyên dương, "))
count=0
while n>0:
    count+=1
    n//=10
print("Số chữ số của n là:", count)
```

##### Bài 16: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ 


```python
n=int(input("Nhập vào một số nguyên dương, "))
even_count=0
odd_count=0
while n>0:
    digit=n%10
    if digit%2==0:
        even_count+=1
    else:
        odd_count+=1
    n//=10
print("Số chữ số chẵn là:", even_count)
print("Số chữ số lẻ là:", odd_count)
```

##### Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 

```python
##### Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 
A=int(input("Nhập vào một số nguyên dương, "))
a, b=1, 1
while b<=A:
    a, b=b, a+b
print("Số trong dãy số fibonacci lớn nhất nhưng không vượt quá A là:", a)
```

##### Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố. 

```python
##### Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố. 
def is_prime(num):
    if num < 2:
        return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            return False
    return True
n = int(input("Nhập vào một số nguyên dương: "))
count = 0
temp = n
while temp > 0:
    digit = temp % 10
    if is_prime(digit):
        count += 1
    temp //= 10
print("Số chữ số nguyên tố trong n là:", count)

```
