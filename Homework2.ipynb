##### Bài 1: Nhập vào 2 cạnh của một hình chữ nhật. In ra màn hình diện tích và chu vi của nó.  

```python
a = float(input("Nhâp cạnh a"))
b = float(input("Nhập cạnh b"))
if a>0 and b>0:
    S = a*b
    P = (a+b)*2
    print("Diện tích hình chữ nhật là: ", S)
    print("Chu vi hình chữ nhật là: ", P)
else:
    print("Cạnh a và b phải lớn hơn 0")
```

##### Bài 3 : Tính tam giác: Nhâp vào ba số a,b,c bất kì. Kiểm tra xem nó có thể là độ dài ba cạnh hay không, nếu  không thì in  ra màn  hình  'Khong la ba  canh cua mot tam  giac'. Ngược lại, kiểm tra xem nó là tam giác gì: Đều, Cân, Vuông, hay Tam giác thường. Tính diện tích, chu vi của tam giác và in ra màn hình. 

```python
a, b, c = map(float, input("Nhập cạnh của tam giác: ").split())
if a>0 and b>0 and c>0:
  if a+b>c and a+c>b and b+c>a:
    if a==b and b==c and c==a:
      print("Tam giác đều")
    elif a==b or b==c or c==a:
      print("Tam giác cân")
    elif a**2+b**2==c**2 or a**2+c**2==b**2 or b**2+c**2==a**2:
      print("Tam giác vuông")
    else:
      print("Tam giác thường")
  else:
    print("Không là ba cạnh của một tam giác")
else:
  print("Không là ba cạnh của một tam giác")
```

##### Bài 5: Viết chương trình giải phương trình bậc 2 (ax2 + bx + c = 0) 


```python
import math
a = float(input("Nhập vào hằng số bậc hai: "))
b = float(input("Nhập vào hằng số bậc nhất: "))
c = float(input("Nhập vào hằng số tự do: "))
if a!=0:
    delta = b**2 - 4*a*c
    if delta < 0:
      print("Phương trình vô nghiệm")
    elif delta == 0:
      x = -b/(2*a)
      print(f"Phương trình có nghiệm kép x = {x}")
    elif delta > 0:
      x1 = (-b+math.sqrt(delta))/(2*a)
      x2 = (-b-math.sqrt(delta))/(2*a)
      print(f"Phương trình có hai nghiệm phân biệt x1 = {x1} và x2 = {x2}")
else:
    if b==0 and c==0:
     print("Phương trình có vô số nghiệm")
    elif b==0 and c!=0:
     print("Phương trình vô nghiệm")
    else:
      x = -c/b
      print(f"Phương trình có nghiệm x = {x}")
```

##### Bài 8: Giải hệ phương trình tuyến tính:

##### Bài 9: Nhập vào thời gian 1 công việc nào đó là x giây. Hãy chuyển đổi và viết ra màn hình số thời gian trên dưới dạng bằng bao nhiêu giờ, bao nhiêu phút, bao nhiêu giây.  

```python
x =float(input("Số giờ người đó làm việc là: "))
gio = x // 3600
gio = int(gio)
phut = (x % 3600)//60
phut = int(phut)
giay = x  - gio*3600- phut*60
giay = int(giay)
print(f"Thời gian người dó làm việc là: {gio} giờ {phut} phút {giay} giây")
```

##### Bài 10: Nhập vào tâm và bán kính của một đường tròn. Sau đó nhập vào một điểm A(x, y) bất kì và kiểm tra xem nó có thuộc đường tròn hay không? 

```python
import math
x = float(input("Nhập tọa độ x của tâm đường tròn"))
y = float(input("Nhập tọa độ y của tâm đường tròn"))
r = float(input("Nhập bán kình đường tròn"))
x0 = float(input("Nhập tọa độ x của A"))
y0 = float(input("Nhập tọa độ y của A"))
d = math.sqrt((x-x0)**2 + (y-y0)**2)
if r>0:
    if d==r and d>r:
      print("A thuộc đường tròn")
    else:
      print("A không thuộc đường tròn")
else:
   print("Bán kính phải lớn hơn 0")
```
