```python
# W4A1: Viết chương trình Python sử dụng một trong 2 loại vòng lặp for hoặc while để tính tổng các số nguyên từ 1 đến n. (n <= 1000)
n = int(input("Nhập một số nguyên dương n (n <= 1000): "))
total = 0
for i in range(1, n + 1):
    total += i
print("Tổng các số nguyên từ 1 đến", n, "là:", total)
```

```python
# W4A2: Viết chương trình nhập vào một số tới khi được một số nguyên dương thì thôi. Kiểm tra số này có phải là số nguyên tố hay không?
while True:
    n = int(input("Nhập một số nguyên dương: "))
    if n > 0:
        break

is_prime = True
if n < 2:
    is_prime = False
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

if is_prime:
    print(n, "là số nguyên tố")
else:
    print(n, "không phải là số nguyên tố")
```

```python
# W4A3: Viết chương trình tính giai thừa của một số nguyên n cho trước (n! = 1×2×…×n) với 0 < n < 100.
n = int(input("Nhập một số nguyên dương n (0 < n < 100): "))
if 0 < n < 100:
    giaithua = 1
    for i in range(1, n + 1):
        giaithua *= i
    print(n, "! =", giaithua)
else:
    print("Số nhập vào không hợp lệ.")
```

```python
# W4A4: Viết chương trình nhập vào một số nguyên n và in ra màn hình số chữ số (trừ dấu) của số đó. Không dùng kiểu dữ liệu string.
n = int(input("Nhập một số nguyên: "))
count = 0
if n < 0:
    n = -n
while n > 0:
    n //= 10
    count += 1
print("Số chữ số của số đã nhập là:", count)
```

```python
# W4A7: Nhập vào số nguyên dương n (n ≥ 2). Hãy in ra ước số nguyên tố lớn nhất của n.
n = int(input("Nhập một số nguyên dương n (n ≥ 2): "))
so_lon_nhat = None
for i in range(2, int(n ** 0.5) + 1):
    while n % i == 0:
        so_lon_nhat = i
        n //= i
if n > 1:
    so_lon_nhat = n
print("Ước số nguyên tố lớn nhất của n là:", so_lon_nhat)
```

```python
# W4A8: Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 
n = int(input("Nhập số nguyên dương n: "))
def reverse(num):
    rev = 0
    while num > 0:
        rev = rev * 10 + num % 10
        num //= 10
    return rev
steps = 0
while True:
    if n == reverse(n):
        break
    n += reverse(n)
    steps += 1
print(f"Số bước: {steps}, giá trị palindrome thu được: {n}")
```

```python
# WA10: Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số nhỏ nhất." 
n = int(input("Nhập vào số nguyên dương n: "))
def collatz_length(x):
    length = 0
    while x != 1:
        if x % 2 == 0:
            x //= 2
        else:
            x = 3 * x + 1
        length += 1
    return length
max_length = 0
result_x = 1
for i in range(1, n + 1):
    length = collatz_length(i)
    if length > max_length:
        max_length = length
        result_x = i
print(f"{result_x} {max_length}")
```

```python
#W4A13: Hai số nguyên dương tạo thành một cặp số thân thiết khi chúng tuân theo quy luật: Số này bằng tổng tất cả các ước của số kia (trừ chính nó) và ngược lại. Ví dụ 220 và 284 là một cặp số thân thiết. 220 ngoài chính nó ra, nó có 11 ước số khác là: 1, 2, 4, 5, 10, 11, 20, 44, 55, 110. Tổng của 11 ước này cho chúng ta 284. 284 ngoài bản thân nó, còn có 5 ước số khác là: 1, 2, 4, 71, 142. Tổng của chúng vừa đúng bằng 220. Nhập hai số từ bàn phím và kiểm tra xem chúng có phải là cặp số thân thiết không. Nếu có, in ra màn hình "true", ngược lại, in ra "false". 
a = int(input("Nhập số nguyên dương a: "))
b = int(input("Nhập số nguyên dương b: "))
sum_a = 0
for i in range(1, a):
    if a % i == 0:
        sum_a += i
sum_b = 0
for i in range(1, b):
    if b % i == 0:
        sum_b += i
if sum_a == b and sum_b == a:
    print("true")
else:
    print("false")
```

```python
#W4A14: Cho hai số m và n là hai số nguyên dương nhập từ bàn phím. Viết chương trình tìm ước chung lớn nhất của hai số m và n. 
m = int(input("Nhập số nguyên dương m: "))
n = int(input("Nhập số nguyên dương n: "))
while n:
    m, n = n, m % n
print("Ước chung lớn nhất của m và n là:", m)
```
