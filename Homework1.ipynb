```python
# Bài 1: Nhập vào số nguyên n, in ra màn hình số gấp đôi của n.
n = int(input("Nhập vào số nguyên n: "))
gấp_đôi = n * 2
print("Số gấp đôi của n là:", gấp_đôi)
```

```python
# Bài 2: Một khu đất hình chữ nhật có chiều rộng a, chiều dài b. Người ra xây một khu vui chơi hình tròn tại vị trí trung tâm khu đất như hình vẽ. Nhập vào a, b. Hãy tính phần diện tích phần đất trống cây xung quanh là bao nhiêu. Biết pi = 3.14, lấy kết quả đến 2 chữ số phần thập phân.
a = float(input("Nhập vào chiều rộng a: "))
b = float(input("Nhập vào chiều dài b: "))
pi = 3.14
bán_kính = min(a, b) / 2
diện_tích_khu_vui_chơi = 3.14 * bán_kính ** 2
diện_tích_khu_đất = a * b
diện_tích_đất_trống = diện_tích_khu_đất - diện_tích_khu_vui_chơi
print("Diện tích phần đất trống cây xung quanh là:", round(diện_tích_đất_trống, 2))
```

```python
# Bài 3: Viết chương trình nhập vào một “kí tự” c, nếu nó là chữ hoa thì in ra chữ thường tương ứng và ngược lại nếu nó là chữ thường thì in ra chữ hoa tương ứng của nó. (xem trước cấu trúc lựa chọn: if...else, ma)
c = input("Nhập vào một kí tự c: ")
if c.isupper():
    print("Chữ thường tương ứng của c là:", c.lower())
elif c.islower():
    print("Chữ hoa tương ứng của c là:", c.upper())
```

```python
# Bài 4: Viết chương trình nhập vào một “kí tự”, Kiểm tra xem c có phải là kí tự chữ cái không (hoa hoặc thường). Nếu đúng thì in ra “{c} là kí tự alphabet”, nếu sai in ra “{c} không phải là kí tự alphabet”.
c = input("Nhập vào một kí tự: ")
if c.isalpha():
    print(f"{c} là kí tự alphabet")
else:
    print(f"{c} không phải là kí tự alphabet")
```

```python
# Bài 5: Nhập vào chữ cái hoa A, in ra chữ cái thường liền trước chữ cái thường tương ứng của nó. A. Xét TH khi c = ‘A’ là đặc biệt. 
c = input("Nhập vào một chữ cái hoa A: ")
if c == 'A':
    print("Chữ cái thường liền trước chữ cái thường tương ứng của A là: z")
elif c.isupper():
    chữ_cái_thường = c.lower()
    chữ_cái_thường_liền_trước = chr(ord(chữ_cái_thường) - 1)
    print(f"Chữ cái thường liền trước chữ cái thường tương ứng của {c} là: {chữ_cái_thường_liền_trước}")
```

```python
# Bài 6: Nhập vào 3 số a, b, c. Kiểm tra xem đây có phải là cạnh của 1 tam giác không, nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó không phải 3 cạnh của tam giác thì in ra “Không phải 3 cạnh của tam giác”.
a = float(input("Nhập vào cạnh a: "))
b = float(input("Nhập vào cạnh b: "))
c = float(input("Nhập vào cạnh c: "))
if a + b > c and a + c > b and b + c > a:
    p = (a + b + c) / 2
    diện_tích = (p * (p - a) * (p - b) * (p - c)) ** 0.5
    print("Diện tích của tam giác là:", round(diện_tích, 1))
else:
    print("Không phải 3 cạnh của tam giác")
```

```python
# Bài 7: Nhập vào một chuỗi gồm các kí tự thường độ dài >=20, in ra kí tự thứ năm và thứ chín của chuỗi này.
chuỗi = input("Nhập vào một chuỗi gồm các kí tự thường độ dài >=20: ")
if len(chuỗi) >= 20:
    print("Kí tự thứ năm của chuỗi là:", chuỗi[4])
    print("Kí tự thứ chín của chuỗi là:", chuỗi[8])
else:
    print("Chuỗi phải có độ dài >=20")
```

##### Bài 8: Nhập vào tên chủ hộ và chỉ số điện kế tháng trước và chỉ số điện kế tháng này, tính tiền điện tháng và cho biết:
##### - Từ 0 – 50kWh: 1.984 đồng/kWh  
##### - Bậc 2: Từ 51 – 100 kWh: 2.050 đồng/kWh  
##### - Bậc 3: Từ 101 – 200 kWh: 2.380 đồng/kWh  
##### - Bậc 4: Từ 201 – 300 kWh: 2.998 đồng/kWh  
##### - Bậc 5: Từ 301 – 400 kWh: 3.350 đồng/kWh  
##### - Bậc 6: Từ 401 trở đi kWh: 3.460 đồng/kWh

```python

tên_chủ_hộ = input("Nhập vào tên chủ hộ: ")
chỉ_số_điện_tháng_trước = int(input("Nhập vào chỉ số điện kế tháng trước: "))
chỉ_số_điện_tháng_này = int(input("Nhập vào chỉ số điện kế tháng này: "))
số_kWh = chỉ_số_điện_tháng_này - chỉ_số_điện_tháng_trước
if số_kWh <= 50:
    tiền_điện = số_kWh * 1.984
elif số_kWh <= 100:
    tiền_điện = 50 * 1.984 + (số_kWh - 50) * 2.050
elif số_kWh <= 200:
    tiền_điện = 50 * 1.984 + 50 * 2.050 + (số_kWh - 100) * 2.380
elif số_kWh <= 300:
    tiền_điện = 50 * 1.984 + 50 * 2.050 + 100 * 2.380 + (số_kWh - 200) * 2.998
elif số_kWh <= 400:
    tiền_điện = 50 * 1.984 + 50 * 2.050 + 100 * 2.380 + 100 * 2.998 + (số_kWh - 300) * 3.350
else:
    tiền_điện = 50 * 1.984 + 50 * 2.050 + 100 * 2.380 + 100 * 2.998 + 100 * 3.350 + (số_kWh - 400) * 3.460
print(f"Chủ hộ {tên_chủ_hộ} phải trả tiền điện là: {tiền_điện:.2f} đồng")
```
