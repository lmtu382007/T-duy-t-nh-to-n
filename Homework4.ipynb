```python
#W4A20 Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 2^k hay không 
n = int(input("Nhập vào số nguyên dương n: "))
if n > 0 and (n & (n - 1)) == 0:
    print(f"{n} là số dạng 2^k.")
else:
    print(f"{n} không phải là số dạng 2^k.")
```

```python
#W4A21 Nhập vào số nguyên dương n, tính tổng các chữ số của n 
n = int(input("Nhập vào số nguyên dương n: "))
tong = 0
while n>0:
    tong += n % 10
    n //= 10
print(f"Tổng các chữ số của n là: {tong}")
```

```python
#W4A22 Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu chữ số lẻ 
n = int(input("Nhập vào số nguyên dương n: "))
chan = 0
le = 0
while n>0:
    so = n%10
    if so % 2 == 0:
        chan+=1
    else:
        le+=1
    n //= 10
print(f"Số chữ số chẵn: {chan}, số chữ số lẻ: {le}")
```

```python
# Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k) lớn nhất nhưng nhỏ hơn n
n = int(input("Nhập vào số nguyên dương n: "))
k = 0
while k * (k + 1) // 2 < n:
    k += 1
k-=1
print(f"Giá trị của k là: {k}")
```

```python
#Nhập vào A, tìm n nhỏ nhất sao cho 1 + 1/2 + 1/3 + 1/4 + ... + 1/n > A 
A = int(input("Nhập vào số thực dương A: "))
n = 0
tong = 0
while tong <= A:
    n+=1
    tong+=1/n
print(f"Giá trị của n là: {n}")
```

```python
#Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong dãy số fibonacci lớn nhất nhưng không vượt quá A 
A = int(input("Nhập vào số nguyên dương A: "))
a, b = 1, 1
while a <= A:
    a, b = b, a + b
print(f"Số trong dãy số fibonacci lớn nhất nhưng không vượt quá A là: {b-a}")
```

```python
# Nhập vào một dãy số nguyên, ngưng nhập khi người dùng nhập -1. Sau khi nhập xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. -1 có thể là min hoặc max
max_so = -1e9
min_so = 1e9
while True:
    n = int(input("Nhập vào số nguyên (nhập -1 để kết thúc): "))
    if n == -1:
        break
    if n > max_so:
        max_so = n
    if n < min_so:
        min_so = n
print(f"Số lớn nhất: {max_so}, số nhỏ nhất: {min_so}")
```

```python
#Viết chương trình nhập vào một số nguyên n và in ra màn hình số chữ số (trừ dấu) của số đó. Không dùng kiểu dữ liệu string. 
n = int(input("Nhập vào một số nguyên: "))
dem = 0
if n < 0:
    n = -n
while n > 0:
    n //= 10
    dem += 1
print(f"Số chữ số của số đó là: {dem}")
```

```python
#Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được. 
n = int(input("Nhập số nguyên dương n: "))
def reverse(num):
    rev = 0
    while num > 0:
        rev = rev * 10 + num % 10
        num //= 10
    return rev
steps = 0
while True:
    if n == reverse(n):
        break
    n += reverse(n)
    steps += 1
print(f"Số bước: {steps}, giá trị palindrome thu được: {n}")
```

```python
#Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số nhỏ nhất." 
n = int(input("Nhập vào số nguyên dương n: "))
def collatz_length(x):
    length = 0
    while x != 1:
        if x % 2 == 0:
            x //= 2
        else:
            x = 3 * x + 1
        length += 1
    return length
max_length = 0
result_x = 1
for i in range(1, n + 1):
    length = collatz_length(i)
    if length > max_length:
        max_length = length
        result_x = i
print(f"{result_x} {max_length}")
```
