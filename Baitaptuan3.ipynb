```python
# W3A1: Nhập vào một số nguyên dương có 3 chữ số. Tìm và in ra màn hình “SỐ NGUYÊN” đảo ngược của n. Không dùng kiểu chuỗi, không dùng hàm. Chú ý: kiểm tra chỉ xử lí các số nguyên dương có đúng 3 chữ số. Các số khác không xử lý. Hoán đổi hai số không sd biến tạm. 
n = int(input("Nhập một số nguyên dương có 3 chữ số: "))
if 100 <= n <= 999:
    a = n % 10
    b = (n // 10) % 10
    c = n // 100
    reversed_n = a * 100 + b * 10 + c
    print("SỐ NGUYÊN đảo ngược của n là:", reversed_n)
else:
    print("Vui lòng nhập một số nguyên dương có đúng 3 chữ số.")
```

```python
# W3A3: Kiểm tra xem một số có phải là lũy thừa của 2 hay không. Sử dụng các toán tử bitwise để xác định xem một số có phải là lũy thừa của 2 hay không. Gợi ý: Đối với lũy thừa của 2, n & (n-1) == 0. 
n = int(input("Nhập một số: "))
if n > 0 and (n & (n - 1)) == 0:
    print(n, "là lũy thừa của 2.")
else:
    print(n, "không phải là lũy thừa của 2.")
```

```python
# W3A4: Viết chương trình chia một số m cho n, kết quả thu được làm tròn xuống
m = int(input("Nhập số m: "))
n = int(input("Nhập số n: "))
if n != 0:
    result = m // n
    print("Kết quả của m chia cho n là:", result)
```

```python
# W3A5: Viết chương trình chia một số m cho n, kết quả thu được làm tròn lên
m = int(input("Nhập số m: "))
n = int(input("Nhập số n: "))
if n != 0:
    result = -(-m // n)  
    print("Kết quả của m chia cho n là:", result)
```

```python
# W3A8: [Longer string] Nhập vào hai chuỗi a, b gồm các chữ cái thường. In ra "True" nếu chuỗi a dài hơn chuỗi b. Ngược lại in ra "False".
a = input("Nhập chuỗi a: ")
b = input("Nhập chuỗi b: ")
if len(a) > len(b):
    print("True")
else:
    print("False")
```

```python
# W3A9: [Triangle] Nhập vào 3 số nguyên dương a, b, c. Kiểm tra xem a, b, c có cấu thành độ dài cạnh của 1 tam giác được không. Nếu a, b, c cấu tạo thành được một tam giác, in ra "Yes". Ngược lại in ra "No".
a = int(input("Nhập số a: "))
b = int(input("Nhập số b: "))
c = int(input("Nhập số c: "))
if a + b > c and a + c > b and b + c > a:
    print("Yes")
else:
    print("No")
```

```python
# W3A10: [Largest number] Nhập vào 4 số nguyên. In ra màn hình số nguyên lớn nhất trong 4 số.
a = int(input("Nhập số a: "))
b = int(input("Nhập số b: "))
c = int(input("Nhập số c: "))
d = int(input("Nhập số d: "))
largest = a
if b > largest:
    largest = b
if c > largest:
    largest = c
if d > largest:
    largest = d
print("Số nguyên lớn nhất là:", largest)
```

```python
# W3A11: [Phân loại tam giác] Nhập vào 3 số nguyên dương a, b, c là độ dài 3 cạnh của một tam giác. Kiểm tra 3 số đã nhập có tạo thành một tam giác hợp lệ hay không. Nếu không in ra "Không phải tam giác". Nếu có, phân loại tam giác đó là "Tam giác đều", "Tam giác cân", hay "Tam giác thường".
a = int(input("Nhập số a: "))
b = int(input("Nhập số b: "))
c = int(input("Nhập số c: "))
if a + b > c and a + c > b and b + c > a:
    if a == b == c:
        print("Tam giác đều")
    elif a == b or a == c or b == c:
        print("Tam giác cân")
    else:
        print("Tam giác thường")
else:
    print("Không phải tam giác")
```

```python
# W3A14: [Giải phương trình bậc nhất] Nhập vào hai số thực a và b của phương trình ax + b = 0. Tìm và in ra nghiệm của phương trình, kết quả làm tròn đến số thập phân thứ 2. In ra "Vô nghiệm" hoặc "Vô số nghiệm" trong trường hợp tương ứng.
a = float(input("Nhập số a: "))
b = float(input("Nhập số b: "))
if a == 0:
    if b == 0:
        print("Vô số nghiệm")
    else:
        print("Vô nghiệm")
else:
    x = -b / a
    print("Nghiệm của phương trình là: ", round(x, 2))
```
